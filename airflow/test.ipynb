```python
import json
import requests
import time

# sido.json 로드
with open('sido.json', 'r', encoding='utf-8') as f:
    sido_data = json.load(f)

# 전체 데이터를 담을 구조
all_data = {
    'provinces': [],
    'cities': [],
    'districts': [],
    'hierarchy': []
}

# 각 시/도별로 크롤링
for sido in sido_data:
    print(f"처리 중: {sido['name']}")
    
    # 시/도 정보 저장
    all_data['provinces'].append(sido)
    
    # 시/군/구 가져오기
    gugun_url = f"https://www.weather.go.kr/w/rest/zone/dong.do?type=CITY&wideCode={sido['code'][:-2]}&cityCode=&keyword=&keywordStart=&keywordEnd="
    gugun_list = requests.get(gugun_url).json()
    
    # 계층 구조용
    sido_hierarchy = {
        **sido,
        'cities': []
    }
    
    for gugun in gugun_list:
        print(f"  - {gugun['name']}")
        
        # 시/군/구 정보 저장
        all_data['cities'].append(gugun)
        
        # 읍/면/동 가져오기
        dong_url = f"https://www.weather.go.kr/w/rest/zone/dong.do?type=DONG&wideCode={sido['code'][:-2]}&cityCode={gugun['code']}&keyword=&keywordStart=&keywordEnd="
        dong_list = requests.get(dong_url).json()
        
        # 읍/면/동 정보 저장
        all_data['districts'].extend(dong_list)
        
        # 계층 구조에 추가
        sido_hierarchy['cities'].append({
            **gugun,
            'districts': dong_list
        })
        
        time.sleep(0.3)  # API 부하 방지
    
    all_data['hierarchy'].append(sido_hierarchy)
    time.sleep(0.5)

# JSON 파일로 저장
with open('all_regions.json', 'w', encoding='utf-8') as f:
    json.dump(all_data, f, ensure_ascii=False, indent=2)

print(f"\n완료!")
print(f"시/도: {len(all_data['provinces'])}개")
print(f"시/군/구: {len(all_data['cities'])}개")
print(f"읍/면/동: {len(all_data['districts'])}개")
```
